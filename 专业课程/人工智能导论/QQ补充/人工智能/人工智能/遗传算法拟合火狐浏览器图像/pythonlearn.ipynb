```python
def sub(x,y):
    return x-y
print(sub(12,8))
print(sub(8,12))
print(sub(y=8,x=12))#key arguments
```

**可变参数**
**如果可变参数后面还有其他参数的话要当作关键字参数去传值**

```python
def girlsschool(age=18, *sex,**name):
    print(age)
    print(sex[2])
    print(len(sex))
    print(name)
girlsschool(14,0,1,1,name='eric')
```

**可变参数允许你传入 0 个或任意个参数，我们知道这些可变参数在函数调用时自动组装为一个 tuple。而关键字参数允许你传入 0 个或任意个含参数名的参数，这些关键字参数在函数内部自动组装为一个 dict。**

```python
def girlsschool(age=18, *sex,**name):
    print(age)
    print(sex[2])
    print(len(sex))
    print(name)
girlsschool(0,1,1,name='eric')
```

# 正则表达式re模块

```python
import re
sample='134pythonwqr43'
result=re.search('python',sample)
print(result)
print(result.group())
```

```python
import re
phonenum='my phonenum is 123-456-789'
phonenumre=re.compile(r'(\d\d\d)-(\d\d\d-\d\d\d)')
res=phonenumre.search(phonenum)
print(res.group(1))
print(res.group(2))
print(res.group())
print(res.groups())
fir,sec=res.groups()
print(fir,sec)
```

如果希望匹配多个字符串其中一个，可以用|

```python
import re
mo=re.compile(r'math|english')
res1=mo.search('chinese,math and english')
res2=mo.search('chinese,english and math')
print(res1.group())
#会优先匹配第一个
print(res2.group())
```
